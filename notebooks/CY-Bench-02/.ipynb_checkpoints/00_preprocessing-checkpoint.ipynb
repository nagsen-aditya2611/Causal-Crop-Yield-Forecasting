```python
#imports and path

from pathlib import Path
import pandas as pd
import numpy as np

BASE = Path("..")          # This path is for local use
RAW = BASE / "data" / "raw" / "CY-Bench"
PROCESSED = BASE / "data" / "processed" / "CY-Bench"
PROCESSED.mkdir(parents=True, exist_ok=True)

CROP = "wheat"
COUNTRY = "IN"
```

```python
#file inventory
crop_dir = RAW / CROP / COUNTRY

files = sorted([p for p in crop_dir.iterdir() if p.suffix.lower() == ".csv"])
pd.DataFrame({"file": [p.name for p in files]})
```

```python
# loading tables
def load_csv(name):
    return pd.read_csv(crop_dir / name)

yield_df = load_csv(f"yield_{CROP}_{COUNTRY}.csv")
soil_df = load_csv(f"soil_{CROP}_{COUNTRY}.csv")
calendar_df = load_csv(f"crop_calendar_{CROP}_{COUNTRY}.csv")
meteo_df = load_csv(f"meteo_{CROP}_{COUNTRY}.csv")
ssm_df = load_csv(f"soil_moisture_{CROP}_{COUNTRY}.csv")
ndvi_df = load_csv(f"ndvi_{CROP}_{COUNTRY}.csv")
fpar_df = load_csv(f"fpar_{CROP}_{COUNTRY}.csv")
```
